```python
import json
from NatGen.src.data_preprocessors.transformations.no_transform import NoTransformation
from NatGen.src.data_preprocessors.transformations.dead_code_inserter import DeadCodeInserter
import re
        
lang = "c"
path="/home/you/codebert/NatGen/parser/languages.so"
```

```python
dead_code = DeadCodeInserter(path, lang)
nt = NoTransformation(path, lang)


def transform_dead_code(code):
    code, meta = dead_code.transform_code(code)
    return code

def no_transform(code):
    code, meta = nt.transform_code(code)
    return code
                
def gen_transform_dead_code(filename, js_all):
    
    with open(filename, 'w') as f:
        for idx,js in enumerate(js_all):
            js['func'] = transform_dead_code(js['func'])
            f.write(json.dumps(js)+'\n')
            
            
def gen_transform_nt(js_all):
    
    ntl = []
   
    for idx,js in enumerate(js_all):
        print(js)
        c = no_transform(js['func'])
        js['func'] = c
        ntl.append(js)
        
    return ntl
```

```python
data_dead_code_devign = [json.loads(line) for line in open('dataset/test_inj1.jsonl', 'r')]
```

```python
diffl = []
samps = 0
for i in range(len(nt)):
    c1 = nt[i]
    c2 = data_dead_code_devign[i]
    if c1 != c2:
        samps += 1
        diffl.append((c1,c2))
```

```python
len(diffl)
```

```python
import difflib as dl
dead_code_mapper = {}
for x in diffl:

    a = x[0]['func'].split(' ')
    b = x[1]['func'].split(' ')

    s = dl.SequenceMatcher(None, a, b)
    s.get_matching_blocks()

    replace_this = ""
    for i in range(len(s.get_matching_blocks())-1):
        m = s.get_matching_blocks()[i]
        m_next =  s.get_matching_blocks()[i+1]
        if m_next.b != (m.b + m.size):
            replace_this = " ".join(b[m.size:m_next.b])


    dead_code_mapper[x[1]['idx']] = replace_this
```

```python
len(dead_code_mapper)
```

```python
def gen_transform_dead_code_appendix_1inj(filename, js_all):
    
    with open(filename, 'w') as f:
        for idx,js in enumerate(js_all):
            if js['idx'] in dead_code_mapper.keys():
                js['dead_code'] = dead_code_mapper[js['idx']]
            else:
                js['dead_code'] = ''
            
            f.write(json.dumps(js)+'\n')

def gen_transform_dead_code_appendix_2inj(filename, js_all):
    
    with open(filename, 'w') as f:
        for idx,js in enumerate(js_all):
            if js['idx'] in dead_code_mapper.keys():
                js['dead_code_2'] = dead_code_mapper[js['idx']]
            else:
                js['dead_code_2'] = ''
            
            f.write(json.dumps(js)+'\n')
            
def gen_transform_dead_code_appendix_3inj(filename, js_all):
    
    with open(filename, 'w') as f:
        for idx,js in enumerate(js_all):
            if js['idx'] in dead_code_mapper.keys():
                js['dead_code_3'] = dead_code_mapper[js['idx']]
            else:
                js['dead_code_3'] = ''
            
            f.write(json.dumps(js)+'\n')
            
def gen_transform_dead_code_appendix_4inj(filename, js_all):
    
    with open(filename, 'w') as f:
        for idx,js in enumerate(js_all):
            if js['idx'] in dead_code_mapper.keys():
                js['dead_code_4'] = dead_code_mapper[js['idx']]
            else:
                js['dead_code_4'] = ''
            
            f.write(json.dumps(js)+'\n')
            
def gen_transform_dead_code_appendix_5inj(filename, js_all):
    
    with open(filename, 'w') as f:
        for idx,js in enumerate(js_all):
            if js['idx'] in dead_code_mapper.keys():
                js['dead_code_5'] = dead_code_mapper[js['idx']]
            else:
                js['dead_code_5'] = ''
            
            f.write(json.dumps(js)+'\n')
```
